```python
# Initialize mappings
emotion_to_core = {}
emotion_to_fine = {}

# Populate the mappings using the emotion_wheel
for core_emotion, subdict in emotion_wheel.items():
    # Map core emotions to themselves
    emotion_to_core[core_emotion] = core_emotion
    emotion_to_fine[core_emotion] = core_emotion
    
    for mid_level, fine_emotions in subdict.items():
        # Map mid-level emotions to their core
        emotion_to_core[mid_level] = core_emotion
        # Map mid-level emotions to themselves
        emotion_to_fine[mid_level] = mid_level
        
        # Map fine-grained emotions to their core and mid-level
        for fine_emotion in fine_emotions:
            emotion_to_core[fine_emotion] = core_emotion
            emotion_to_fine[fine_emotion] = mid_level
```

```python
import spacy
import re
import contractions
# Load the spaCy model once (outside the function)
language_model = "en_core_web_trf"
nlp = spacy.load(language_model)

def preprocess_text(text, stop_words=False, remove_punctuation=False, keep_emotion_punct=True):
    """
    Preprocess text for emotion classification.

    Args:
        text (str): Input text.
        stop_words (bool): Whether to remove stopwords.
        remove_punctuation (bool): Whether to remove ALL punctuation.
        keep_emotion_punct (bool): Whether to keep emotion-related punctuation (!, ?).

    Returns:
        tokens (list): List of tokens.
        lemmas (list): List of lemmas.
    """
    # Step 1: Make everything lowercase
    text = text.lower()

    # Step 2: Handle contractions (e.g., "can't" → "cannot")
    text = contractions.fix(text)

    # Step 3: Remove special characters
    if keep_emotion_punct:
        # Keep emotion-related punctuation (!, ?)
        text = re.sub(r"[^a-zA-ZăâîșțĂÂÎȘȚ0-9\s!?]", "", text)
    else:
        # Remove all punctuation
        text = re.sub(r"[^a-zA-ZăâîșțĂÂÎȘȚ0-9\s]", "", text)

    # Step 4: Tokenization
    doc = nlp(text)
    tokens = []
    lemmas = []
    for token in doc:
        # Skip stopwords if enabled
        if stop_words and token.is_stop:
            continue
        # Skip punctuation if enabled
        if remove_punctuation and token.is_punct:
            continue
        # Append token and lemma
        tokens.append(token.text)
        lemmas.append(token.lemma_)

    return tokens, lemmas

text = "I bought strawberries! They were amazing, but I can't believe how expensive they were."

# Keep emotion punctuation (!, ?) and handle contractions
tokens, lemmas = preprocess_text(
    text, 
    stop_words=True, 
    remove_punctuation=False,  # Keep punctuation
    keep_emotion_punct=True
)

print("Tokens:", tokens)
print("Lemmas:", lemmas)
```

```python
# Define preprocessing function (updated to return cleaned text)
def preprocess_text(text, remove_stopwords=True):
    # Handle contractions (e.g., "can't" → "cannot")
    text = contractions.fix(text)
    
    # Remove user mentions (@username)
    text = re.sub(r"@\w+", "[USER]", text)
    
    # Remove URLs
    text = re.sub(r"http\S+", "[URL]", text)
    
    # Normalize numbers
    text = re.sub(r"\d+", "[NUM]", text)
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove special characters except basic punctuation and emotion cues
    text = re.sub(r"[^a-zA-Z0-9\s.!?,;:'\-]", "", text)
    
    # Tokenize and lemmatize with spaCy
    doc = nlp(text)
    tokens = []
    for token in doc:
        # Skip stopwords if enabled (punctuation is never a stopword)
        if remove_stopwords and token.is_stop:
            continue
        # Keep punctuation tokens (e.g., "!", "?")
        tokens.append(token.text)
    # Join lemmas into a cleaned text string
    cleaned_text = " ".join(tokens)
    
    return cleaned_text
```

```python
# Step 1: Create corpus and calculate global term frequencies
corpus = ' '.join(core_dataset['cleaned_text'])
word_freq_dict = Counter(corpus.split())

# Step 2: Get min and max frequencies for scaling
all_frequencies = list(word_freq_dict.values())
min_freq = min(all_frequencies) if all_frequencies else 0
max_freq = max(all_frequencies) if all_frequencies else 0

# Step 3: Function to replace words with scaled frequencies (0-1)
def replace_words_with_scaled_frequency(sentence, freq_dict, min_f, max_f):
    words = sentence.split()
    scaled_freq = []
    for word in words:
        freq = freq_dict.get(word, 0)  # Use 0 if word not in vocabulary
        # Min-Max scaling (handle division by zero if all frequencies are equal)
        if max_f == min_f:
            scaled = 0.5  # Arbitrary value if all frequencies are the same
        else:
            scaled = (freq - min_f) / (max_f - min_f)
        scaled_freq.append(round(scaled, 4))  # Round for readability
    return scaled_freq

# Step 4: Apply the function to create a new column
core_dataset['tf_scaled'] = core_dataset['cleaned_text'].apply(
    lambda x: replace_words_with_scaled_frequency(x, word_freq_dict, min_freq, max_freq)
)

# Display the updated DataFrame
core_dataset
```

```python
import torch
import torch.nn as nn
from transformers import BertModel, BertTokenizer

# Load BERT tokenizer
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# Define a simple model
class EmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained("bert-base-uncased")
        self.dropout = nn.Dropout(0.1)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)
    
    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids, attention_mask=attention_mask)
        pooled_output = outputs.last_hidden_state[:, 0, :]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

# Initialize model
num_classes = len(label_encoder.classes_)
model = EmotionClassifier(num_classes=num_classes)  # Replace with your model
```

```python
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader, Dataset
import torch.optim as optim

class EmotionDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
    
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = self.texts.iloc[idx]
        label = self.labels.iloc[idx]  # Now an integer
        
        encoding = self.tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )
        
        return {
            "input_ids": encoding["input_ids"].flatten(),
            "attention_mask": encoding["attention_mask"].flatten(),
            "label": torch.tensor(label, dtype=torch.long)  # Works with integers
        }

# 2. Split dataset (now with numerical labels)
train_df, test_df = train_test_split(core_dataset, test_size=0.2, stratify=core_dataset["emotion"])

# 3. Initialize Dataset and DataLoader (no other changes needed)
train_dataset = EmotionDataset(train_df["cleaned_text"], train_df["emotion"], tokenizer)
test_dataset = EmotionDataset(test_df["cleaned_text"], test_df["emotion"], tokenizer)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16)

# 4. Training loop remains the same

# Optimizer & Loss
optimizer = optim.AdamW(model.parameters(), lr=2e-5)
criterion = nn.CrossEntropyLoss()

# Training loop (simplified)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

for epoch in range(3):  # Start with 3 epochs
    model.train()
    for batch in train_loader:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["label"].to(device)
        
        optimizer.zero_grad()
        outputs = model(input_ids, attention_mask)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

```python
# Section 6: Evaluation & Results
# --------------------------------
from sklearn.metrics import classification_report

model.eval()
y_true = []
y_pred = []

with torch.no_grad():
    for batch in test_loader:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["label"].cpu().numpy()
        
        outputs = model(input_ids, attention_mask)
        preds = torch.argmax(outputs, dim=1).cpu().numpy()
        
        y_true.extend(labels)
        y_pred.extend(preds)

# Map emotion indices to labels
emotion_labels = core_dataset["emotion"].unique().tolist()
print(classification_report(y_true, y_pred, target_names=emotion_labels))
```
